```python
SYSTEM_PROMPT = """
You are an expert learning roadmap designer.

Your job is to create personalized learning roadmaps
for people who want to learn a specific domain or skill.

The roadmap must be:

- Practical
- Realistic
- Structured
- Easy to follow
- Personalized to the learner's level
- Appropriate for the available learning time
- Ordered from fundamentals to advanced concepts

Always include:

1. Roadmap Overview
2. Learning Phases
3. Timeline
4. Topics to Learn
5. Practice Activities
6. Practical Projects
7. Tools and Technologies
8. Expected Outcomes
9. Next Steps

Important rules:

- Do not overload the learner with unnecessary topics.
- Prioritize the most important skills.
- Start with fundamentals when the learner is a beginner.
- Avoid advanced topics if they are not necessary.
- Adjust the depth of the roadmap according to the available time.
- Make the roadmap practical and career-oriented.
"""
```

```python
def generate_roadmap(domain, level, time):

    if not domain.strip():
        return "❌ Please enter a domain or field."

    user_prompt = f"""
Create a personalized learning roadmap for the following learner.

Domain:
{domain}

Current Skill Level:
{level}

Available Learning Time:
{time}

Create a realistic roadmap that matches this learner.

Explain:

- What they should learn
- In what order they should learn it
- How much time they should spend on each phase
- What they should practice
- What projects they should build
- What tools they should learn

Format the answer using clear headings,
bullet points, and tables where useful.
"""

    try:

        response = client.responses.create(
            model="openai/gpt-oss-20b",
            instructions=SYSTEM_PROMPT,
            input=user_prompt
        )

        return response.output_text

    except Exception as e:

        return f"❌ Error: {str(e)}"
```

```python
result = generate_roadmap(
    "Data Analysis",
    "Beginner",
    "3 months"
)

print(result)
```

```python
import gradio as gr

with gr.Blocks(title="AI Learning Roadmap Generator") as app:

    gr.Markdown("""
    # 🚀 AI Learning Roadmap Generator

    Create a personalized learning roadmap using AI.
    """)

    domain = gr.Textbox(
        label="📚 Domain / Field",
        placeholder="Example: Data Analysis"
    )

    level = gr.Dropdown(
        choices=[
            "Beginner",
            "Intermediate",
            "Advanced"
        ],
        value="Beginner",
        label="🎯 Skill Level"
    )

    time = gr.Dropdown(
        choices=[
            "1 Month",
            "3 Months",
            "6 Months",
            "1 Year"
        ],
        value="3 Months",
        label="⏱️ Learning Time"
    )

    generate_button = gr.Button(
        "🚀 Generate Roadmap"
    )

    roadmap_output = gr.Markdown(
        label="Your Learning Roadmap"
    )

    generate_button.click(
        fn=generate_roadmap,
        inputs=[
            domain,
            level,
            time
        ],
        outputs=roadmap_output
    )

app.launch(share=True)
```
